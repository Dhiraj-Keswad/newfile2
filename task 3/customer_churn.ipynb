```python
a.columns=(a.columns.str.strip().str.lower().str.replace(" ","_"))

a.rename(columns={'customerid':'customer_id', 'age':'age', 'gender':'gender', 'tenure':'tenure', 'usagefrequency':"usage_frequency",
       'supportcalls':'support_calls', 'paymentdelay':"payment_delay", 'subscriptiontype':'subscription_type', 'contractlength':'contract_length',
       'totalspend':'total_spend', 'lastinteraction':'last_interaction', 'churn':'churn'},inplace=True)
a.columns.str.strip().str.lower()
a.columns
```

Changing Datatypes

```python
a.dtypes
```

```python
# changing from float column tointeger column 
integer_column=['customer_id', 'age', 'tenure', 'usage_frequency',
       'support_calls', 'payment_delay', 'last_interaction', 'churn','total_spend']
for col in integer_column:
    print(col)
    a[col]=pd.to_numeric(a[col],errors="coerce")

for col in integer_column:
    a[col]=a[col].round().astype("int64")

text_clmn=[ 'gender','subscription_type',
'contract_length']   
for col in text_clmn:
    a[col]=a[col].astype("string").str.strip().str.title()
print(a.dtypes)    
```

```python
# replacing churn values from 0,1 to no,yes respectively
a["churn"]=a["churn"].astype("string")
a["churn"]=a["churn"].replace({"1":"Yes","0":"no"})
a.tail()
```

```python
a["churn"].value_counts()
```

Detecting Outliers using IQR

```python
for col in a.columns:
    if a[col].dtype=="int64":
        print(col)
    else :
        0
        
```

```python
numeric_columns=("customer_id","age","tenure","usage_frequency","support_calls","payment_delay","total_spend","last_interaction")
for col in numeric_columns:
    Q1=a[col].quantile(0.25)
    Q3=a[col].quantile(0.75)
    IQR=Q3-Q1
    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    outliers= a[(a[col]<lower) |
                 (a[col]>upper)]
    print(col,"outliers:",len(outliers))
```

```python
# Feature engineering
# creating a new column for spend_per_tenure and high_support_flag
a["spend_per_tenure"]=((a["total_spend"]/ a["tenure"]).round(2).astype(float))
a["high_support_flag"]=(a["support_calls"]>=7).astype("string")
a["high_support_flag"]=a["high_support_flag"].replace({"True":"Yes","False":"no"})
```
